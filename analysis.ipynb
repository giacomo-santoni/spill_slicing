```python
import ROOT
```

```python
f1name = "files/SAND_2/sand-spill-events.0.overlay.edep.root"
#f2 = ROOT.TFile("files/sand-spill-events.1.overlay.edep.root")
tname = "EDepSimEvents"

f = ROOT.TFile(f1name)
tree = f.Get(tname)
# t.Print()

df = ROOT.RDataFrame(tname, f1name)
```

```python
x_coord = []
y_coord = []
z_coord = []
t_coord = []
for i in range(tree.GetEntries()):
    def1 = df.Define("x", f"Primaries.Position[{i}].X()") \
             .Define("y", f"Primaries.Position[{i}].Y()") \
             .Define("z", f"Primaries.Position[{i}].Z()") \
             .Define("t", f"Primaries.Position[{i}].T()")
    x = def1.Take["double"]("x").GetValue()
    y = def1.Take["double"]("y").GetValue()
    z = def1.Take["double"]("z").GetValue()
    time = def1.Take["double"]("t").GetValue()
    x_coord.append(x)
    y_coord.append(y)
    z_coord.append(z)
    t_coord.append(time)
```
